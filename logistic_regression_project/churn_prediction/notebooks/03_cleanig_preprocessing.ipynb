### Step 4 - Remove Irrelevant Features

Not every feature in a dataset contributes to predicting the target variable.

Some columns are unique identifiers, some contain geographical information with little predictive value, while others introduce **data leakage** because they contain information that is only available after the customer has already churned.

To build a reliable machine learning model, these features should be removed before further analysis.

The following columns will be removed:

- **CustomerID** – Unique identifier for each customer.
- **Count** – Constant value with no predictive information.
- **Country** – Contains the same value for every customer.
- **State** – Contains only one state and provides no useful variation.
- **City** – High-cardinality feature that adds unnecessary complexity.
- **Zip Code** – Geographic identifier rather than customer behavior.
- **Lat Long** – Combined geographic coordinates.
- **Latitude** – Geographic coordinate.
- **Longitude** – Geographic coordinate.
- **Churn Score** – Generated after churn and causes data leakage.
- **CLTV** – Customer Lifetime Value calculated using future information.
- **Churn Reason** – Only available for customers who already churned and contains over 73% missing values.

Removing these features simplifies the dataset and ensures that the model learns only from information available before churn occurs.

```python
columns_to_drop = [
    "CustomerID",
    "Count",
    "Country",
    "State",
    "City",
    "Zip Code",
    "Lat Long",
    "Latitude",
    "Longitude",
    "Churn Score",
    "CLTV",
    "Churn Reason"
]

df_clean.drop(columns=columns_to_drop, inplace=True)
```

### Observation

The unnecessary and leakage-prone features have been successfully removed.

The dataset now contains only the features that are relevant for predicting customer churn, making it more suitable for further preprocessing and model development.
