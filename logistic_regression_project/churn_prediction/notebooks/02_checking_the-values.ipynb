### Step 6 — Calculate the Percentage of Missing Values

While the total number of missing values is useful, calculating their percentage provides better insight into the severity of missing data.

Columns with a very high percentage of missing values may need to be removed, whereas columns with only a few missing values can often be treated using suitable imputation techniques.

```python
missing_values = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Percentage (%)": (df_clean.isnull().sum() / len(df_clean)) * 100
})

missing_values[missing_values["Missing Values"] > 0].sort_values(
    by="Percentage (%)",
    ascending=False
)
```

### Observation

The missing value report helps identify which features require preprocessing.

Based on the percentage of missing values, we can decide whether to remove a feature or apply an appropriate imputation technique during data cleaning.
