### Step 3: Missing Value Analysis

Missing values occur when no data is available for a particular feature.

Before deciding how to handle them, we first need to determine:

- Which columns contain missing values.
- How many missing values exist.
- The percentage of missing values in each column.

This helps us choose an appropriate strategy for handling incomplete data.

```python
missing_count = df.isnull().sum()

missing_percentage = (
    missing_count / len(df) * 100
).round(2)

missing_df = pd.DataFrame({
    "Column": missing_count.index,
    "Missing Count": missing_count.values,
    "Missing Percentage (%)": missing_percentage.values
})

missing_df = missing_df[
    missing_df["Missing Count"] > 0
].sort_values(
    by="Missing Percentage (%)",
    ascending=False
)

missing_df
```

### Observations

The table above shows there is no any missing values 

For each column, we can observe:

- The total number of missing record are zero 
- The percentage of missing values relative to the entire dataset is also zero 

These observations help determine whether missing values should be removed or imputed.

```python
if missing_df.empty:
    print("No missing values found in the dataset.")
else:
    print("Missing values detected.")
```

### Missing Value Strategy

Since the dataset does not contain missing values, no imputation or deletion was required.

Therefore, the original data remains unchanged.

### Step 5: Duplicate Value Analysis

Duplicate rows represent repeated observations in the dataset.

Keeping duplicate records may introduce bias during model training because the same information is counted multiple times.

Therefore, we check whether duplicate rows exist before proceeding.

```python
duplicate_rows = df.duplicated().sum()

print(f"Duplicate Rows : {duplicate_rows}")
```

```python
if duplicate_rows > 0:
    df.drop_duplicates(inplace=True)

    print("Duplicate rows removed successfully.")
else:
    print("No duplicate rows found.")
```

### step 7: check the outlier in this data set 
- we check the outlier by using mean median 
- the outlier can be problem to unusuall pattern in the model 



```python
mean_salary = df['salary_in_usd'].mean()
median_salary = df['salary_in_usd'].median()

print(f"Mean Salary: {mean_salary}")
print(f"Median Salary: {median_salary}")
```

#### the outlier in the raw salary going to further like 3 milllion plus may be because of currency conversation 
- so we remove the salary column and salary_currency
- because of raw salary table have the massive numbers

```python
df = df.drop(columns=['salary', 'salary_currency'])
```

### Step 8: Verify the Cleaned Dataset

After completing the cleaning process, we verify the dataset again to ensure:

- No missing values remain.
- No duplicate records remain.
- The dataset is ready for preprocessing.

```python
print("Remaining Missing Values")
print(df.isnull().sum().sum())

print()

print("Remaining Duplicate Rows")
print(df.duplicated().sum())

print()

print("Final Dataset Shape")
print(df.shape)
```

* there is still the 800,0000 masive salary that can be outlier like 1754 or the extreme outlier 
* so we remove the row that contains salary above 330,000 

```python
Q1 = df['salary_in_usd'].quantile(0.25)
Q3 = df['salary_in_usd'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR

print(f"Original row count: {len(df)}")
print(f"Dropping salaries above: ${upper_bound:,.2f}")

# 3. Filter the dataframe to KEEP only rows below the upper bound
df_final = df[df['salary_in_usd'] <= upper_bound]

print(f"New row count: {len(df_final)}")
print(f"Total outliers removed: {len(df) - len(df_final)}")
```

### Step 9: Save Clean Dataset

The cleaned dataset is saved inside the **clean_data** directory.

This dataset will be used in the next notebook for preprocessing.

```python
df_final.to_csv(
    "D:/machine learning/machine_learning/linear_regression_project/salary_prediction/data/clean_data/cleaned_salaries.csv",
    index=False
)

print("Clean dataset saved successfully.")
```
